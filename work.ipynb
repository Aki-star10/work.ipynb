```python
# %% 1. Write to a file
file = open("students.txt", "w")
file.write("John\n")
file.write("Mary\n")
file.write("Peter\n")
file.close()
print("File created")
```

```python
# %% 2. Read the whole file
file = open("students.txt", "r")
content = file.read()
print(content)
file.close()
```

```python
# %% 3. Read one line
file = open("students.txt", "r")
print(file.readline())
file.close()
```

```python
# %% 4. Read all lines as a list
file = open("students.txt", "r")
lines = file.readlines()
print(lines)
file.close()
```

```python

# %% 5. Loop through the file
file = open("students.txt", "r")
for line in file:
    print(line.strip())
file.close()
```

```python
# %% 6. Add a new name (append)
file = open("students.txt", "a")
file.write("Grace\n")
file.close()
```

```python
file = open("students.txt", "r")
print(file.read())
file.close()
```

```python
# %% 8. Using "with" (closes the file for you)
with open("fruits.txt", "w") as file:
    file.write("banana\n")
    file.write("apple\n")
    file.write("orange\n")
```

```python

with open("fruits.txt", "r") as file:
    print(file.read())
```

```python
# ---------- 15. EXCEPTION HANDLING ----------
print("--- 15. try-except ---")
try:
    print(10 / 0)
except:
    print("Oops! You cannot divide by zero")
 
print("The program did not crash. Done!")
```

```python

# ---------- 1. IF-ELSE ----------
print("--- 1. if-else ---")
age = 20
 
if age >= 18:
    print("You are an adult")
else:
    print("You are a minor")
```

```python
# ---------- 2. NESTED IF ----------
print("--- 2. nested if ---")
number = 10
 
if number > 0:
    if number % 2 == 0:
        print("Positive and even")
    else:
        print("Positive and odd")
else:
    print("Zero or negative")
```

```python
# ---------- 3. FOR LOOP ----------
print("--- 3. for loop ---")
for i in range(1, 6):
    print(i)
```

```python
# ---------- 4. WHILE LOOP ----------
print("--- 4. while loop ---")
count = 1
while count <= 5:
    print(count)
    count = count + 1
```

```python
# ---------- 5. FUNCTION ----------
print("--- 5. function ---")
 
def say_hello(name):
    print("Hello", name)
 
say_hello("Amina")
say_hello("John")
```

```python

# ---------- 6. FUNCTION THAT RETURNS A VALUE ----------
print("--- 6. function with return ---")
 
def add(a, b):
    return a + b
 
answer = add(3, 4)
print("3 + 4 =", answer)
```

```python

# ---------- 7. RECURSION ----------
# A function that calls itself. This one counts down.
print("--- 7. recursion ---")
 
def countdown(n):
    if n == 0:
        print("Go!")
    else:
        print(n)
        countdown(n - 1)
 
countdown(3)
```

```python
# ---------- 8. LAMBDA ----------
# A lambda is a short one-line function.
print("--- 8. lambda ---")
double = lambda x: x * 2
print(double(5))
```

```python
# ---------- 9. LIST ----------
print("--- 9. list ---")
fruits = ["apple", "banana", "mango"]
fruits.append("orange")
print(fruits)
print("First fruit:", fruits[0])
```

```python
# ---------- 10. TUPLE ----------
# Like a list, but you cannot change it.
print("--- 10. tuple ---")
days = ("Mon", "Tue", "Wed")
print(days)
print("First day:", days[0])
```

```python
# ---------- 11. SET ----------
# A set removes duplicates.
print("--- 11. set ---")
numbers = {1, 2, 2, 3, 3, 3}
print(numbers)
```

```python
# ---------- 12. DICTIONARY ----------
# Stores data as key: value pairs.
print("--- 12. dictionary ---")
student = {"name": "Amina", "age": 20}
print(student["name"])
student["age"] = 21
print(student)
```
